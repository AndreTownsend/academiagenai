```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
import warnings
warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────
# 1. DATASETS
# ─────────────────────────────────────────────
years_hist = np.arange(2015, 2026)

# Language School Industry ($B global revenue)
lang_school_rev = [60.2, 63.1, 66.5, 70.2, 73.8, 58.1, 62.4, 66.9, 68.5, 70.1, 67.3]

# AI Translation/Language Tool market ($B)
ai_lang_market  = [0.5, 0.8, 1.3, 2.1, 3.4, 5.2, 7.8, 12.1, 18.6, 26.4, 38.2]

# Gen AI Adoption in Higher Ed (% of institutions)
genai_edu_adopt = [2, 4, 6, 9, 15, 22, 35, 52, 68, 81, 91]

# International student enrollment (M students globally)
intl_students   = [4.5, 4.8, 5.1, 5.4, 5.6, 4.1, 4.6, 5.0, 5.3, 5.6, 5.9]

# AI Translation accuracy (% BLEU score proxy)
ai_accuracy     = [65, 68, 72, 76, 80, 84, 88, 91, 93, 95, 96.5]

# Online language platform users (M) — Duolingo, Babbel, etc.
online_lang_users = [15, 25, 45, 80, 130, 180, 240, 350, 480, 590, 680]

df = pd.DataFrame({
    "Year": years_hist,
    "LangSchool_Rev_B": lang_school_rev,
    "AI_Lang_Market_B": ai_lang_market,
    "GenAI_EduAdopt_Pct": genai_edu_adopt,
    "Intl_Students_M": intl_students,
    "AI_Accuracy_Pct": ai_accuracy,
    "Online_LangUsers_M": online_lang_users,
})

# ─────────────────────────────────────────────
# 2. FORECAST 2026–2035
# ─────────────────────────────────────────────
forecast_years = np.arange(2026, 2036)

def poly_forecast(series, years_hist, years_fore, degree=2, clip=None):
    X = years_hist.reshape(-1, 1)
    poly = PolynomialFeatures(degree)
    Xp = poly.fit_transform(X)
    model = LinearRegression().fit(Xp, series)
    Xf = poly.transform(years_fore.reshape(-1, 1))
    preds = model.predict(Xf)
    if clip: preds = np.clip(preds, clip[0], clip[1])
    return preds

fore_lang   = poly_forecast(np.array(lang_school_rev), years_hist, forecast_years, clip=(10, 100))
fore_ai     = poly_forecast(np.array(ai_lang_market),  years_hist, forecast_years, clip=(0, 500))
fore_adopt  = poly_forecast(np.array(genai_edu_adopt), years_hist, forecast_years, clip=(70, 100))
fore_online = poly_forecast(np.array(online_lang_users), years_hist, forecast_years, clip=(0, 3000))
fore_acc    = poly_forecast(np.array(ai_accuracy),     years_hist, forecast_years, clip=(90, 99.5))

fore_df = pd.DataFrame({
    "Year": forecast_years,
    "LangSchool_Rev_B": fore_lang,
    "AI_Lang_Market_B": fore_ai,
    "GenAI_EduAdopt_Pct": fore_adopt,
    "Online_LangUsers_M": fore_online,
    "AI_Accuracy_Pct": fore_acc,
})

full_df = pd.concat([df, fore_df], ignore_index=True).fillna(method="ffill")
full_df["Period"] = full_df["Year"].apply(lambda y: "Historical" if y <= 2025 else "Forecast")

# ─────────────────────────────────────────────
# 3. LANGUAGE SCHOOL SEGMENTS — Threat Assessment
# ─────────────────────────────────────────────
segments = pd.DataFrame({
    "Segment":          ["ESL / EFL Programs", "Business English", "Test Prep (IELTS/TOEFL)",
                          "Conversational Classes", "Academic Writing", "Cultural Immersion",
                          "Early Childhood Language", "University Lang. Depts.", "Corporate Training"],
    "AI_Disruption_Risk": [85, 78, 70, 62, 80, 20, 25, 72, 75],
    "Current_Market_B":   [18.5, 9.2, 7.4, 8.1, 4.2, 3.8, 5.5, 6.2, 4.4],
    "Projected_Loss_Pct_2035": [42, 38, 35, 28, 45, 5, 8, 40, 39],
    "Survivability_Score":      [35, 42, 50, 58, 30, 95, 90, 45, 42],
})

# ─────────────────────────────────────────────
# 4. AI TOOLS DISRUPTING LANGUAGE EDUCATION
# ─────────────────────────────────────────────
ai_tools = pd.DataFrame({
    "Tool":         ["ChatGPT / Claude", "Google Translate", "DeepL", "Duolingo AI",
                      "Microsoft Translator", "Pimsleur AI", "Speechify", "Elsa Speak"],
    "Capability_Pct": [95, 90, 92, 75, 88, 70, 72, 80],
    "Users_M":        [200, 500, 30, 600, 120, 5, 20, 15],
    "Cost_vs_School": [-99, -100, -99, -95, -100, -70, -85, -80],  # % cheaper vs school
})

# ─────────────────────────────────────────────
# 5. VISUALIZATION
# ─────────────────────────────────────────────
plt.style.use("dark_background")
fig = plt.figure(figsize=(22, 16), facecolor="#0c0a1e")
fig.suptitle("GEN AI IN ACADEMIA  |  Will Language Schools Survive the AI Revolution?",
             fontsize=20, fontweight="bold", color="#f59e0b", y=0.98)

# 5.1 Revenue comparison: Language Schools vs AI Market
ax1 = fig.add_subplot(3, 3, (1, 2))
hist = full_df[full_df.Period == "Historical"]
fore = full_df[full_df.Period == "Forecast"]
ax1.plot(hist.Year, hist.LangSchool_Rev_B, "b-o", lw=2, ms=5, label="Language School Revenue ($B) - Historical")
ax1.plot(fore.Year, fore.LangSchool_Rev_B, "b--", lw=2, label="Language School Revenue ($B) - Forecast")
ax1.plot(hist.Year, hist.AI_Lang_Market_B, "r-o", lw=2, ms=5, label="AI Language Market ($B) - Historical")
ax1.plot(fore.Year, fore.AI_Lang_Market_B, "r--", lw=2, label="AI Language Market ($B) - Forecast")
ax1.axvline(2025.5, color="yellow", ls=":", lw=1.5, alpha=0.7, label="Forecast Start")
# Crossover annotation
ax1.annotate("🚨 AI Market crosses\nSchool Revenue ~2026",
             xy=(2026, 67), xytext=(2019, 85),
             arrowprops=dict(arrowstyle="->", color="yellow"),
             fontsize=8, color="yellow")
ax1.set_title("Language School Revenue vs. AI Language Tool Market", color="white", fontsize=12)
ax1.legend(fontsize=8, loc="upper left")
ax1.set_facecolor("#111827")
ax1.tick_params(colors="white")
for spine in ax1.spines.values(): spine.set_edgecolor("#333")

# 5.2 GenAI Adoption in Higher Ed
ax2 = fig.add_subplot(3, 3, 3)
ax2.fill_between(full_df.Year, full_df.GenAI_EduAdopt_Pct, alpha=0.3, color="#f59e0b")
ax2.plot(full_df.Year, full_df.GenAI_EduAdopt_Pct, color="#f59e0b", lw=2.5)
ax2.axvline(2025.5, color="white", ls=":", lw=1.5, alpha=0.7)
ax2.set_ylim(0, 105)
ax2.set_title("Gen AI Adoption in Higher Education (%)", color="white", fontsize=11)
ax2.set_facecolor("#111827")
ax2.tick_params(colors="white")
for spine in ax2.spines.values(): spine.set_edgecolor("#333")

# 5.3 Segment Disruption Risk
ax3 = fig.add_subplot(3, 3, (4, 5))
seg_sorted = segments.sort_values("AI_Disruption_Risk", ascending=False)
colors_seg = plt.cm.RdYlGn_r(seg_sorted.AI_Disruption_Risk.values / 100)
bars3 = ax3.barh(seg_sorted.Segment, seg_sorted.AI_Disruption_Risk,
                  color=colors_seg, edgecolor="none", height=0.65)
ax3.axvline(50, color="white", lw=1, ls="--", alpha=0.5)
ax3.set_title("AI Disruption Risk by Language School Segment (%)", color="white", fontsize=12)
ax3.set_facecolor("#111827")
ax3.set_xlim(0, 105)
ax3.tick_params(colors="white")
for spine in ax3.spines.values(): spine.set_edgecolor("#333")
for bar, val in zip(bars3, seg_sorted.AI_Disruption_Risk):
    ax3.text(val + 0.5, bar.get_y() + bar.get_height()/2, f"{val}%",
             va="center", color="white", fontsize=8)
ax3.text(51, -0.8, "High Risk →", color="white", fontsize=7, alpha=0.6)

# 5.4 AI Translation Accuracy
ax4 = fig.add_subplot(3, 3, 6)
ax4.plot(full_df.Year, full_df.AI_Accuracy_Pct, color="#44ff88", lw=2.5)
ax4.fill_between(full_df.Year, 60, full_df.AI_Accuracy_Pct, alpha=0.2, color="#44ff88")
ax4.axhline(90, color="white", lw=1, ls="--", alpha=0.5)
ax4.text(2015.5, 91, "Human Parity Threshold", color="white", fontsize=8, alpha=0.7)
ax4.axvline(2025.5, color="yellow", ls=":", lw=1.5, alpha=0.7)
ax4.set_title("AI Translation Accuracy (%) — BLEU Score Equivalent", color="white", fontsize=11)
ax4.set_facecolor("#111827")
ax4.set_ylim(60, 102)
ax4.tick_params(colors="white")
for spine in ax4.spines.values(): spine.set_edgecolor("#333")

# 5.5 Online Language Platform Growth
ax5 = fig.add_subplot(3, 3, (7, 8))
ax5.fill_between(full_df.Year, full_df.Online_LangUsers_M, alpha=0.3, color="#a78bfa")
ax5.plot(full_df.Year, full_df.Online_LangUsers_M, color="#a78bfa", lw=2.5)
ax5.axvline(2025.5, color="yellow", ls=":", lw=1.5, alpha=0.7)
ax5.set_title("Online Language Platform Users (M) — Duolingo, Babbel & AI Apps", color="white", fontsize=11)
ax5.set_facecolor("#111827")
ax5.tick_params(colors="white")
for spine in ax5.spines.values(): spine.set_edgecolor("#333")

# 5.6 Survivability Matrix
ax6 = fig.add_subplot(3, 3, 9)
ax6.set_facecolor("#111827")
ax6.axis("off")
verdict_text = [
    ("🎓 VERDICT", "#f59e0b", 13, "bold"),
    ("", "white", 6, "normal"),
    ("❌ HIGH RISK (Score < 50)", "#ff4444", 10, "bold"),
    ("   ESL Programs, Academic Writing,", "white", 9, "normal"),
    ("   University Depts, Corporate Training", "white", 9, "normal"),
    ("", "white", 5, "normal"),
    ("⚠️  MODERATE (Score 50–70)", "#ffdd44", 10, "bold"),
    ("   Business English, Test Prep,", "white", 9, "normal"),
    ("   Conversational Classes", "white", 9, "normal"),
    ("", "white", 5, "normal"),
    ("✅  SAFE (Score 70+)", "#44ff88", 10, "bold"),
    ("   Cultural Immersion Programs,", "white", 9, "normal"),
    ("   Early Childhood Language Schools", "white", 9, "normal"),
    ("", "white", 5, "normal"),
    ("💡 Schools that pivot to human-", "white", 9, "normal"),
    ("   centered, experiential learning", "white", 9, "normal"),
    ("   will outlast the AI wave.", "#f59e0b", 9, "normal"),
]
y = 0.97
for text, color, size, weight in verdict_text:
    ax6.text(0.05, y, text, color=color, fontsize=size, fontweight=weight,
             transform=ax6.transAxes, va="top")
    y -= 0.065

fig.text(0.5, 0.01, "Andre Townsend, MBA  |  Data Analytics Portfolio  |  Project 4: Gen AI in Academia",
         ha="center", color="#888", fontsize=9)

plt.tight_layout(rect=[0, 0.03, 1, 0.97])
plt.savefig("project4_genai_academia.png", dpi=150, bbox_inches="tight", facecolor="#0c0a1e")
plt.show()
print("✅ Project 4 chart saved.")

full_df.to_csv("project4_genai_academia.csv", index=False)
segments.to_csv("project4_segments.csv", index=False)
ai_tools.to_csv("project4_ai_tools.csv", index=False)
print("✅ CSVs exported for Power BI.")
print("\n📊 2035 Forecast:")
row35 = full_df[full_df.Year == 2035].iloc[0]
print(f"   Language School Market: ${row35['LangSchool_Rev_B']:.1f}B")
print(f"   AI Language Market:     ${row35['AI_Lang_Market_B']:.1f}B")
print(f"   AI Translation Acc:     {row35['AI_Accuracy_Pct']:.1f}%")
print(f"   Online Platform Users:  {row35['Online_LangUsers_M']:.0f}M")
```
